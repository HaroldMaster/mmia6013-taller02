**Pregunta 1:** Con tu tabla de la Parte 2.b delante. Hasta la revisión anterior del evaluador, un sistema que acertaba todas las preguntas respondibles reportaba un Hit Rate de 0,70 con la plantilla tal como se distribuía. ¿Qué medía ese número, qué cambia al sacar las negativas del denominador, y por qué la métrica correcta para ellas es otra? Responde con tus cifras, no de memoria.

```python
import csv

with open("../resultados.csv", encoding="utf-8") as f:
    filas = [r for r in csv.DictReader(f) if r["k"] == "5"]

aciertos = sum(r["hit"] == "True" for r in filas)
respondibles = sum(r["respondible"] == "True" for r in filas)

print("aciertos:", aciertos)
print("Hit Rate sobre respondibles:", aciertos, "/", respondibles, "=", round(aciertos / respondibles, 3))
print("Hit Rate con las negativas en el denominador:", aciertos, "/", len(filas), "=", round(aciertos / len(filas), 3))
print("máximo posible con las negativas dentro:", respondibles, "/", len(filas), "=", round(respondibles / len(filas), 3))
```

Con las 8 preguntas que tienen respuesta, el Hit Rate dio 0,750 (6 de 8) y el MRR 0,583, igual con k=3 y con k=5. Antes, el evaluador también contaba las 2 preguntas negativas. Con los mismos resultados el Hit Rate habría bajado a 0,60 (6 de 10), y aunque el sistema acertara todas, lo máximo sería 0,80 (8 de 10). O sea, ese número no solo medía si el buscador encontraba el paper correcto, sino también cuántas negativas tenía el golden set. Al sacarlas, el Hit Rate mide solo la búsqueda. Las negativas no tienen un paper correcto, así que no tiene sentido medirlas con Hit Rate; lo que importa es si el sistema se abstiene. La abstención correcta dio 1,000, pero hay que verla junto con la indebida (0,250 con k=3 y 0,286 con k=5), porque un sistema que nunca responde también sacaría 1,000.

Escribe una pregunta global sobre tu corpus, explica por qué tu RAG vectorial no puede contestarla bien, y qué haría GraphRAG distinto con ella.

Una pregunta global sobre este corpus sería: "¿Cuáles son los temas principales de los 16 papers y cómo se relacionan?". El RAG que se armó no puede responderla bien porque solo trae los 3 o 5 fragmentos más parecidos a la pregunta (de 948) y el modelo responde solo con eso. Esta pregunta no se parece a ningún fragmento en especial, y la respuesta está repartida en todos los papers. Edge et al. GraphRAG hace algo distinto: primero usa un LLM para armar un grafo con los conceptos y cómo se relacionan, lo separa en grupos y hace un resumen de cada grupo. Después cada resumen da una respuesta parcial y al final se juntan en una sola respuesta.
